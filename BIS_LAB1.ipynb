```python
# Different Representations of a TSP Solution

tour = [1, 4, 8, 2, 5, 3, 6, 7]

print("Original Tour:")
print(tour)

# 1. Path Representation
path = tour

print("\n1. Path Representation:")
print(path)


# 2. Adjacency Representation
adjacency = {}

for i in range(len(tour)):
    current = tour[i]
    next_city = tour[(i + 1) % len(tour)]
    adjacency[current] = next_city

print("\n2. Adjacency Representation:")
print(adjacency)


# 3. Ordinal Representation
cities = sorted(tour)
remaining = cities.copy()
ordinal = []

for city in tour:
    position = remaining.index(city)
    ordinal.append(position + 1)
    remaining.pop(position)

print("\n3. Ordinal Representation:")
print(ordinal)
```

```python
# Modified Cycle Crossover (CX2)
# For TSP

def cx2(parent1, parent2):
    n = len(parent1)

    child1 = []
    child2 = []

    # Start with first city of parent2
    city = parent2[0]
    child1.append(city)

    # Find its position in parent1
    pos = parent1.index(city)

    # Get corresponding city from parent2
    city = parent2[pos]
    child2.append(city)

    # Generate children
    while len(child1) < n:

        # Find position of last city of child1 in parent1
        pos1 = parent1.index(child1[-1])

        # Get city at same position from parent2
        next_city1 = parent2[pos1]

        if next_city1 not in child1:
            child1.append(next_city1)

        # Find position of last city of child2 in parent1
        pos2 = parent1.index(child2[-1])

        # Get city at same position from parent2
        next_city2 = parent2[pos2]

        if next_city2 not in child2:
            child2.append(next_city2)

        # If no new city is added, fill remaining cities
        if len(child1) < n and next_city1 in child1:
            for city in parent1:
                if city not in child1:
                    child1.append(city)
                    break

        if len(child2) < n and next_city2 in child2:
            for city in parent2:
                if city not in child2:
                    child2.append(city)
                    break

    return child1, child2


# Example parents
parent1 = [1, 2, 3, 4, 5, 6, 7, 8]
parent2 = [2, 7, 5, 8, 4, 1, 6, 3]

child1, child2 = cx2(parent1, parent2)

print("Parent 1:", parent1)
print("Parent 2:", parent2)

print("\nChild 1:", child1)
print("Child 2:", child2)
```
